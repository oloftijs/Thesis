# QEggRoll Integer MLP Training Demo
This notebook demonstrates how to load and train the pure-integer IntMLP module programmatically.

```python
import jax
import jax.numpy as jnp
import numpy as np

from qeggroll.nn import IntMLP
from qeggroll.optim import QEggRoll
from qeggroll.data import make_xor_dataset
from qeggroll.utils import simple_es_tree_key
```

```python
rng = np.random.default_rng(42)
train_x, train_y = make_xor_dataset(1024, rng)

model_key, es_key = jax.random.split(jax.random.key(0))
frozen_params, params, scan_map, es_map = IntMLP.rand_init(
    model_key, in_dim=2, out_dim=2, hidden_dim=32, n_layer=1, dtype="int8"
)
es_tree_key = simple_es_tree_key(params, es_key, scan_map)

frozen_noiser, noiser = QEggRoll.init_noiser(params, sigma_shift=1, update_threshold=1)

print("Model and optimizer initialized.")
```
